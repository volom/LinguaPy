```python
import pandas as pd
import sqlite3
import plotly.express as px
```

```python
db_path = r"db_repo/linguapy.db"
```

```python
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# w_stat
cursor.execute("SELECT * FROM w_stat")
rows = cursor.fetchall()
column_names = [description[0] for description in cursor.description]
w_stat = pd.DataFrame(rows, columns=column_names)

# w_dictionary
cursor.execute("SELECT * FROM w_dictionary")
rows = cursor.fetchall()
column_names = [description[0] for description in cursor.description]
w_dictionary = pd.DataFrame(rows, columns=column_names)

# w_translated
cursor.execute("SELECT * FROM w_translated")
rows = cursor.fetchall()
column_names = [description[0] for description in cursor.description]
w_translated = pd.DataFrame(rows, columns=column_names)
```

```python
w_stat.head(3)
```

```python
grouped_data = w_stat.groupby('count_learned')['word_eng'].count().reset_index()
grouped_data.rename(columns={'word_eng': 'word_count'}, inplace=True)

fig = px.bar(
    grouped_data,
    x='count_learned',
    y='word_count',
    title='Word Count by Count Learned',
    labels={'count_learned': 'Count Learned', 'word_count': 'Word Count'},
    text='word_count'  # Show values on the bars
)

fig.update_traces(textposition='outside')
fig.update_layout(
    xaxis=dict(title='Count Learned'),
    yaxis=dict(title='Word Count'),
    bargap=0.2
)

# Show the chart
fig.show()
```

```python
w_translated['time_added'] = pd.to_datetime(w_translated['time_added'])
daily_counts = w_translated.groupby(w_translated['time_added'].dt.date).size().reset_index(name='count')
daily_counts.rename(columns={'time_added': 'date'}, inplace=True)
fig = px.line(daily_counts, x='date', y='count', title='Daily Row Count Dynamics', labels={'count': 'Row Count', 'date': 'Date'})
fig.show()
```
